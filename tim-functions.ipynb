#### Example 4: Mean Function
Given a list of numbers, find their mean.

````python
# my_mean([1, 2, 3]) -> 2

def my_mean(data):
    n = len(data)
    total = 0
    for item in data:
        total += item
    return total / n
````

````python
my_mean([2, 3, 4])
````

````python
my_mean([3, 1, 6, 2, 1, 6])
````

### Vowel Counting Function
That vowel counting problem was a doozy! Good thing we only had to solve it once. Let's bottle it up into a function so we can use it whenever we want!

````python
# BONUS: Write a function that does this (given a string input)

def vowel_count(text):

    n_vowels = 0
    for char in text:
        if char.lower() in 'aeiou':
            n_vowels += 1

    return n_vowels
````

````python
vowel_count('bananas foster')
````

## FizzBuzz Decomposition

````python
def div_by(n, d):
    return n % d == 0

def fizzbuzz(n):
    for i in range(1, n + 1):
        if div_by(i, 3) and div_by(i, 5):
            print("FizzBuzz")
        elif div_by(i, 3):
            print("Fizz")
        elif div_by(i, 5):
            print("Buzz")
        else:
            print(i)
````

````python
div_by(6, 3)
````

````python
div_by(7, 3)
````

````python
fizzbuzz(15)
````
