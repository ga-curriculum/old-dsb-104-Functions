#### Example 4: Mean Function
Given a list of numbers, find their mean.

````python
def my_mean(nums):
    total = sum(nums)
    n = len(nums)
    return total / n
````

````python
my_mean([85, 92, 97])
````

### Vowel Counting Function
That vowel counting problem was a doozy! Good thing we only had to solve it once. Let's bottle it up into a function so we can use it whenever we want!

````python
def count_vowels(input_str):
    n_vowels = 0
    for char in input_str.lower():
        if char in 'aeiou':
            n_vowels += 1
            
    return n_vowels
````

````python
count_vowels("hello world")
````

### You Try: Pig Latin (another tough one!)
Let's write a function that converts **one single word** into Pig Latin. The (simplified) rules of Pig Latin are as follows:

> Take the first letter of a word and move it to the end and add _-ay_ as a suffix. If that letter was a vowel, add _-way_ isnstead. Only do this for words that are 3 letters or more.

You may assume the whole word is lowercase.

````python
def to_pig_latin_word(word):
    if len(word) <= 2:
        return word
    else:
        first_letter = word[0]
        if first_letter in 'aeiou':
            suffix = 'way'
        else:
            suffix = 'ay'
        return word[1:] + first_letter + suffix
````

````python
to_pig_latin_word('python')
````

````python
to_pig_latin_word('avocado')
````

````python
to_pig_latin_word('is')
````

## Next:
Write a function that converts **an entire sentence** to Pig Latin. You may assume the whole sentence is lowercase and that there is no punctuation.

````python
def to_pig_latin(sentence):
    words = sentence.split(' ')
    pig_words = []
    for word in words:
        pig_words.append(to_pig_latin_word(word))
    return ' '.join(pig_words)
````

````python
to_pig_latin('so long and thanks for all the fish')
````

## Variable Scope
The "scope" of a variable is where it can be accessed. Variables can be seen in functions, but variables defined in functions aren't accessible outside of them.

_(There are ways to bypass these rules, but they generally should never be used!)_

````python
a = "Outside the function!"

def f():
    print(a)
    b = "Inside the function!"
    
f()    
# print(b) # <-- Error! b is not defined.
````

## FizzBuzz Decomposition

````python
def div_by(n, d):
    return n % d == 0

def fizzbuzz(n):
    result = []
    for i in range(1, n + 1):
        if div_by(i, 3) and div_by(i, 5):
            result.append("FizzBuzz")
        elif div_by(i, 3):
            result.append("Fizz")
        elif div_by(i, 5):
            result.append("Buzz")
        else:
            result.append(str(i))
            
    return result
````
